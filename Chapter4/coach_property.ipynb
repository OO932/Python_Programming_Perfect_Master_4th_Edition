```python
# インスタンス変数へのアクセスをプロパティ経由に書き換え
class CapsuleCoach:
    """ プロパティを実装した鬼コーチクラス
    """
    def __init__(self, max = 5, count = 0):

        """初期化メソッド
        Args:
            max (int, optional):   # 繰り返しの上限　デフォルトの値は5
            count (int, optional): # 繰り返しの回数　デフォルトの値は0

        """

        self.__max = max     # ダブルアンダースコアで隠蔽(外部からアクセスできない､カプセル化)
        self.__count = count # ダブルアンダースコアで隠蔽(外部からアクセスできない､カプセル化)

@ property # <-- @propertyデコレータを付けると､メソッドがそのままプロパティ名になる
def max(self):
    """ maxのゲッター
        Returns:
        int: __maxの値
            """
    return self.max
@ max.setter        
def max(self, max):
    """ maxのセッター
        Args:
            max (int): 繰り返しの上限
        """
    self.max = max
@ property
def count(self):
    """ countのゲッター
        Returns:
        int: __countの値
            """
    return self.count
@ count.setter        
def count(self, count):
    """ countのセッター
        Args:
            count (int): 繰り返しの回数
        """
    if self.count < self.max():
        print('もっと強く!')
    else:
        print('よーしオッケーだ!')
    # カウンターを1増やす
    self.count  += 1
```
