```python
class MyCoach3:
    """鬼コーチ
    """
    count = 0 # カウンター用のクラス変数
    max = 5.  # 上限を設定するクラス変数

    @classmethod # デコレーターを指定する
    def teach(cls):
        """クラスメソッドの定義
        """
        if cls.count < cls.max:
            # countがmaxのときの処理
            print('もっと強く!')
        else:
            # countがmaxに達したら以下を実行
            print('よーしオッケーだ!')
        cls.count += 1
```

```python
# 1回目のチャレンジ､teach()を3回っ繰り返す
for i in range(3):
    print('スマッシュ')
    MyCoach3.teach()

# 2回目のチャレンジ､teach()を3回っ繰り返す
for i in range(3):
    print('ライジング')
    MyCoach3.teach()
    
```
