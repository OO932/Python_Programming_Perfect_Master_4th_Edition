```python
class MyCoach2:
    """鬼コーチクラス
    """
    count = 0 # カウンター用のクラス変数
    max = 5.  # 上限を設定するクラス変数

    def Teach(self):
        """熱血指導メソッド
        """
        if self.__class__.count < self.__class__.max:
            # countがmax未満のときの処理
            print('もっと強く!')
        else:
            # countがmaxに達したら以下を実行
            print('よーしオッケーだ!')
        self.__class__.count += 1
```

```python
# 1回目のチャレンジ､Teach()を3回繰り返す
my1 = MyCoach2()
for i in range(3):
    print('スマッシュ')
    my1.teach()

# 2回目のチャレンジ､Teach()を4回繰り返す
my2 = MyCoach2()
for i in range(3):
    print('ライジング')
    my2.teach()
```
