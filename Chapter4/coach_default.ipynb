```python
class DefaultCoach:
    """デフォルト値で初期化される鬼コーチクラス
    """
    # パラメータmaxとcountのデフォルト値を設定
    def __init__ (self, max = 5, count = 0):
        """初期化メソッド

        Args:
            max (int, optional):   # 繰り返しの上限　デフォルトの値は5
            count (int, optional): # 繰り返しの回数　デフォルトの値は0
        """
        # 引数が渡されない場合はデフォルト値で変数を初期化
        # 引数が渡されたら引数の値で変数を初期化
        self.max = max
        self.count = count

    def teach(self):
        """熱血指導メソッド
        """
        if self.count < self.max:
            print('もっと強く!')
        else:
            print('よーしオッケーだ!')
        # カウンターを1増やす
        self.count += 1
```

```python
df = DefaultCoach() # DefaultCoachをインスタンス化 
df.count = 1        # countの初期値を1にする
for i in range(5):
    print('スマッシュ!')
    df.teach() 
```
